# Tokenization

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

max_words = 10000

tokenizer = Tokenizer(num_words = max_words,oov_token="<unk>")
tokenizer.fit_on_texts(X_train[0])


train_sequence = tokenizer.texts_to_sequences(X_train[0])
test_sequence = tokenizer.texts_to_sequences(X_test[0])

 
```

```python
# padding the text - 50
max_padding = 50

padded_train_sequence = pad_sequences(
    train_sequence, maxlen=max_words, padding="post",truncating='post'
)

padded_test_sequence = pad_sequences(
    test_sequence, maxlen=max_words, padding="post",truncating='post'
)

padded_train_sequence
```
